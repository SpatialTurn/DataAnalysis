## Roads as a graph

To a computer, a road map is a **graph**:

- **Nodes** are intersections and dead ends.
- **Edges** are the road segments that connect them. Each edge knows its length in meters.

Once roads are a graph, we can ask questions like "what is the shortest drive from A to B?"
or "how many minutes is each home from the nearest grocery store?"

## Step 1: Import

````python
# osmnx comes installed in the SpatialTurn GIS kernel, so there is nothing to install.

import osmnx as ox
import networkx as nx
import matplotlib.pyplot as plt
````

## Step 2: Download the road network

`network_type="drive"` keeps only roads that cars can use. Other options are `"walk"` and `"bike"`.
This can take up to a minute.

````python
place = "West Lafayette, Indiana, USA"
G = ox.graph_from_place(place, network_type="drive")
````

## Step 3: How big is the network?

````python
print("Nodes (intersections):", G.number_of_nodes())
print("Edges (road segments):", G.number_of_edges())

stats = ox.basic_stats(G)
print(f"Average street segment: {stats['street_length_avg']:.0f} meters")
print(f"Total street length: {stats['street_length_total'] / 1000:.0f} km")
````

## Step 4: Plot the network

````python
fig, ax = ox.plot_graph(G, node_size=5, edge_color="gray", bgcolor="white")
````

## Step 5: Look at nodes and edges as tables

`graph_to_gdfs` splits the graph into two GeoDataFrames, which you already know how to use.

````python
nodes, edges = ox.graph_to_gdfs(G)
nodes[["x", "y", "street_count"]].head()
````

`x` is longitude, `y` is latitude, and `street_count` is how many roads meet at that intersection.

````python
edges[["name", "highway", "length"]].head()
````

`highway` is the road type (residential, primary, and so on) and `length` is in meters.

````python
ax = edges.plot(figsize=(8, 8), linewidth=0.6, color="gray")
nodes.plot(ax=ax, markersize=3, color="red")
ax.set_title("Edges (gray) and nodes (red)")
plt.show()
````
